# 1B: Fit a model in Python

In this notebook you will generate some data, fit a straight line to it and measure the error. Run each cell in order.

## 1. Make some data

We invent a relationship between study hours and test score and add random noise, as real data always has.

````python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
hours = np.linspace(0, 10, 30)
scores = 2.5 * hours + 4 + rng.normal(0, 2.0, size=hours.size)

print(f"We have {hours.size} observations.")
````

## 2. Fit the model

`np.polyfit` finds the slope and intercept that minimise the mean squared error.

````python
slope, intercept = np.polyfit(hours, scores, deg=1)
print(f"Learned model: score = {slope:.2f} x hours + {intercept:.2f}")
````

## 3. Measure the error

````python
predictions = slope * hours + intercept
mse = np.mean((scores - predictions) ** 2)
print(f"Mean squared error: {mse:.2f}")
````

## 4. Look at the result

````python
plt.scatter(hours, scores, label="Observed data")
plt.plot(hours, predictions, color="red", label="Fitted line")
plt.xlabel("Hours studied")
plt.ylabel("Test score")
plt.legend()
plt.show()
````
